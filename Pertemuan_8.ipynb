**Aktivitas 6F: Tabel Perhitungan Total Sementara**

| Elemen | Total sementara |
|---|---:|
| 80 | 80 |
| 65 | 145 |
| 90 | 235 |
| 75 | 310 |
| 70 | 380 |

```python
# === AKTIVITAS 6F: POLA AKUMULATOR ===

skor = [80, 65, 90, 75, 70]

total = 0
jumlah_lulus = 0
for nilai in skor:
    total = total + nilai
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total sementara:", total)

rata_rata = total / len(skor)
print("Total     :", total)
print("Rata-rata :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

**Hasil Pengamatan**

Berdasarkan perhitungan, total akhir dari lima nilai adalah 380 dengan rata-rata 76. Jumlah skor yang bernilai 75 atau lebih adalah 3 dari 5 data. Perulangan for digunakan untuk menjumlahkan setiap nilai, sedangkan percabangan if digunakan untuk menghitung skor yang memenuhi kondisi. Nilai total diperbarui setiap kali perulangan dijalankan.

```python
# === AKTIVITAS 6G: GANTI UKURAN JENDELA ===

suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1 karena kita melihat i - 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 5 hari: jendela = 5 nilai yang berakhir di indeks i
rata_5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata_5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata_5)
print(len(suhu), "nilai ->", len(rata_5), "rata-rata")
```

**Hasil Pengamatan**

Jumlah rata-rata yang diperoleh ada 3 karena dari 7 data suhu, setiap perhitungan menggunakan 5 data secara berurutan. Setelah satu rata-rata dihitung, jendela bergeser satu data ke depan untuk menghitung rata-rata berikutnya. Jadi, dari 7 data hanya bisa diperoleh 3 rata-rata bergerak 5 hari.

```python
# === AKTIVITAS 6H: JELASKAN PENURUNANNYA ===

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

hari = list(range(1, 31))
jam_operasi = [24] * 30
jam_operasi[16] = 18

NAMA_HARI = ["Min", "Sen", "Sel", "Rab", "Kam", "Jum", "Sab"]

# Memeriksa hari dengan jam operasional kurang dari 24 jam
for i in range(len(energi)):
    if jam_operasi[i] < 24:
        print("Hari", hari[i], ":", jam_operasi[i], "jam,", energi[i], "MWh")

# Memisahkan energi hari kerja dan akhir pekan
kerja = []
libur = []

for i in range(len(energi)):
    if NAMA_HARI[i % 7] == "Sab" or NAMA_HARI[i % 7] == "Min":
        libur.append(energi[i])
    else:
        kerja.append(energi[i])

print("Rata-rata hari kerja :", round(sum(kerja) / len(kerja)), "MWh")
print("Rata-rata akhir pekan:", round(sum(libur) / len(libur)), "MWh")

# Menghitung penurunan energi hari ke-17
perubahan = (energi[16] - energi[15]) / energi[15] * 100

print("Penurunan energi hari ke-17:", round(perubahan, 1), "%")
```

**Jawaban Aktivitas 6H**

Berdasarkan hasil pengamatan, energi pada hari ke-17 mengalami penurunan sebesar 41,1% dibandingkan hari ke-16. Jam operasional pada hari tersebut juga hanya 18 jam, sedangkan data awal menunjukkan 24 jam untuk hari lainnya.

Tiga alasan operasional yang mungkin menyebabkan penurunan tersebut adalah:

2. Pengurangan jam operasional. Energi yang dihasilkan mungkin lebih sedikit karena pembangkit hanya beroperasi selama 18 jam. Data yang perlu diperiksa adalah catatan jam operasional dan waktu berhentinya pembangkit.
3. Kegiatan pemeliharaan atau perbaikan. Sebagian peralatan mungkin sedang menjalani pemeliharaan sehingga produksi energi berkurang. Data yang perlu diperiksa adalah jadwal pemeliharaan dan laporan pekerjaan teknisi.
Hasil Pengamatan

Dari hasil perhitungan, penurunan energi pada hari ke-17 mencapai 41,1%. Salah satu faktor yang terlihat adalah berkurangnya jam operasional dari 24 jam menjadi 18 jam. Namun, untuk memastikan penyebab sebenarnya, perlu dilakukan pemeriksaan terhadap catatan gangguan mesin dan kegiatan pemeliharaan.

```python
# === AKTIVITAS 6I: ARRAY ANDA SENDIRI ===

suhu = [28, 31, 30, 33, 29, 27, 32]

# 1. Menampilkan 3 hari pertama dan 2 hari terakhir
print("3 hari pertama:", suhu[:3])
print("2 hari terakhir:", suhu[-2:])

# 2. Menambahkan dan mengeluarkan satu data
suhu.append(34)
print("Setelah append:", suhu)

data_keluar = suhu.pop()
print("Data yang dikeluarkan:", data_keluar)
print("Setelah pop:", suhu)

# 3. Mengurutkan data tanpa mengubah list asli
print("Data terurut:", sorted(suhu))
print("List asli:", suhu)

# 4. Menghitung rata-rata dan mencari suhu tertinggi
rata_rata = sum(suhu) / len(suhu)
suhu_tertinggi = max(suhu)
hari_tertinggi = suhu.index(suhu_tertinggi) + 1

print("Rata-rata suhu:", rata_rata)
print("Suhu tertinggi:", suhu_tertinggi, "°C")
print("Suhu tertinggi terdapat pada hari ke-", hari_tertinggi)

# 5. Membuat list suhu di atas rata-rata
suhu_di_atas_rata = []

for nilai in suhu:
    if nilai > rata_rata:
        suhu_di_atas_rata.append(nilai)

print("Suhu di atas rata-rata:", suhu_di_atas_rata)
```

**Hasil Pengamatan**

Berdasarkan program yang dijalankan, tiga data pertama adalah 28, 31, dan 30, sedangkan dua data terakhir adalah 27 dan 32.

Penambahan data menggunakan append() berhasil dilakukan, kemudian data terakhir dikeluarkan kembali menggunakan pop(). Fungsi sorted() juga berhasil mengurutkan data tanpa mengubah list asli.

Dari 7 data suhu, diperoleh rata-rata sebesar 30 °C. Suhu tertinggi adalah 33 °C yang terdapat pada hari ke-4. Sementara itu, suhu yang berada di atas rata-rata adalah 31, 33, dan 32 °C.

```python
# === PEMANASAN: LIMA LATIHAN KECIL ===

skor = [80, 65, 90, 75, 70, 85]

# 1. Mencetak skor pertama dan terakhir
print("Skor pertama:", skor[0])
print("Skor terakhir:", skor[-1])

# 2. Menghitung rata-rata
rata_rata = sum(skor) / len(skor)
print("Rata-rata:", rata_rata)

# 3. Menghitung skor di atas rata-rata
jumlah_di_atas = 0
for nilai in skor:
    if nilai > rata_rata:
        jumlah_di_atas += 1

print("Skor di atas rata-rata:", jumlah_di_atas)

# 4. Mencari posisi skor terendah
skor_terendah = min(skor)
posisi_terendah = skor.index(skor_terendah) + 1

print("Posisi skor terendah:", posisi_terendah)

# 5. Membuat list baru dengan setiap skor ditambah 5
skor_baru = []
for nilai in skor:
    skor_baru.append(nilai + 5)

print("Skor setelah ditambah 5:", skor_baru)
```

```python
# === LATIHAN PRAKTIKUM: LAPORAN PRODUKSI BULANAN ===

energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,
]

# 1. Menghitung total dan rata-rata energi
total = sum(energi)
rata_rata = total / len(energi)

print("Total energi:", total, "MWh")
print("Rata-rata energi:", round(rata_rata, 1), "MWh/hari")

# 2. Mencari tiga hari tertinggi dan terendah
data_hari = []

for i in range(len(energi)):
    data_hari.append((i + 1, energi[i]))

tertinggi = sorted(data_hari, key=lambda x: x[1], reverse=True)[:3]
terendah = sorted(data_hari, key=lambda x: x[1])[:3]

print("\nTiga hari dengan energi tertinggi:")
for hari, nilai in tertinggi:
    print("Hari", hari, ":", nilai, "MWh")

print("\nTiga hari dengan energi terendah:")
for hari, nilai in terendah:
    print("Hari", hari, ":", nilai, "MWh")

# 3. Menghitung jumlah hari di bawah target
target = 5800
jumlah_di_bawah = 0

for nilai in energi:
    if nilai < target:
        jumlah_di_bawah += 1

print("\nTarget energi:", target, "MWh")
print("Jumlah hari di bawah target:", jumlah_di_bawah, "hari")

# 4. Memeriksa rata-rata bergerak 7 hari
batas = 0.97 * rata_rata

print("\nHari dengan rata-rata bergerak di bawah 97% rata-rata bulanan:")

for i in range(6, len(energi)):
    jendela = energi[i - 6 : i + 1]
    rata_7 = sum(jendela) / 7

    if rata_7 < batas:
        print("Hari", i + 1, ":", round(rata_7, 1), "MWh/hari")

# 5. Menghitung total energi per minggu
print("\nTotal energi per minggu:")

for i in range(0, len(energi), 7):
    minggu = energi[i : i + 7]
    nomor_minggu = i // 7 + 1

    print(
        "Minggu", nomor_minggu, ":",
        sum(minggu), "MWh"
    )
```
