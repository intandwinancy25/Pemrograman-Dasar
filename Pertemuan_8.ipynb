**Prediksi Aktivitas 6A: Tebak Indeks**

Sebelum menjalankan program, saya memprediksi hasil dari empat perintah `print()` sebagai berikut:

1. `suhu[1]` = `31`
2. `suhu[-2]` = `33`
3. `suhu[len(suhu) - 1]` = `29`
4. `suhu[2] + suhu[3]` = `63`

```python
# === AKTIVITAS 6A: TEBAK INDEKS ===

suhu = [28, 31, 30, 33, 29]

print(suhu[1])
print(suhu[-2])
print(suhu[len(suhu) - 1])
print(suhu[2] + suhu[3])
```

**Hasil Perbandingan:**

Hasil yang ditampilkan setelah program dijalankan sesuai dengan prediksi sebelumnya.

**Hal yang Membuat Terkejut:**

Penggunaan indeks negatif `-2` dapat mengambil elemen kedua dari belakang. Selain itu, `len(suhu) - 1` dapat digunakan untuk mendapatkan indeks elemen terakhir dalam list.

```python
# === AKTIVITAS 6B: TEBAK LIST AKHIRNYA ===

angka = [10, 20, 30]

angka.append(40)
angka.insert(1, 15)
angka.remove(30)

print(angka)
print(len(angka))
```

```python
# === AKTIVITAS 6C: CARI BUG-NYA ===

harga = [12, 15, 9]

print("Jumlah elemen:", len(harga))

try:
    print(harga[len(harga)])
except IndexError as error:
    print("IndexError:", error)

print("Cara benar:", harga[len(harga) - 1])
print("Cara sederhana:", harga[-1])
```

**Prediksi Aktivitas 6D**

Sebelum menjalankan program, saya memprediksi hasil sebagai berikut:

1. `suhu[6:12:2]` = `[63, 69, 72]`
2. `len(suhu[::3])` = `8`
3. `suhu[-1] - suhu[0]` = `1`
4. `sorted(suhu)[0]` = `59`

```python
# === AKTIVITAS 6D: TEBAK DULU, BARU JALANKAN ===

suhu = [
    61, 60, 60, 59, 59, 60,   # pukul 00.00–05.00
    63, 66, 69, 71, 72, 73,   # pukul 06.00–11.00
    74, 74, 75, 74, 73, 72,   # pukul 12.00–17.00
    70, 68, 66, 64, 63, 62    # pukul 18.00–23.00
]

# Menampilkan hasil prediksi
print("Hasil suhu[6:12:2]       :", suhu[6:12:2])
print("Jumlah suhu[::3]         :", len(suhu[::3]))
print("Selisih suhu terakhir dan pertama:", suhu[-1] - suhu[0])
print("Suhu terendah            :", sorted(suhu)[0])
```
