**Nomor 1: Membuat Fungsi KPI**

```python
# Nama: Intan Dwinancy Aprilia Nainggolan
# NIM: 4232501035

# Data Unit 1 selama 30 hari
energi = [5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
          6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
          5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062]

batubara = [2998, 3258, 3250, 3263, 3249, 3246, 2993, 2983, 3258, 3258,
            3253, 3249, 3244, 2996, 3009, 3257, 1950, 3234, 3261, 3262,
            3015, 2995, 3260, 3275, 3248, 3264, 3267, 2998, 3008, 3240]

jam_operasi = [24] * 30
jam_operasi[16] = 18

# Data teknis Unit 1
daya_unit1 = 330
nilai_kalor_unit1 = 4400


# Fungsi 1: Heat Rate
def heat_rate(batubara_t, energi_mwh, nilai_kalor):
    """Menghitung heat rate dalam satuan kcal/kWh."""
    if energi_mwh <= 0:
        raise ValueError("Energi harus lebih dari nol.")
    return batubara_t * nilai_kalor / energi_mwh


# Fungsi 2: Efisiensi Termal
def efisiensi_termal(batubara_t, energi_mwh, nilai_kalor):
    """Menghitung efisiensi termal dalam persen."""
    hr = heat_rate(batubara_t, energi_mwh, nilai_kalor)
    return 860 / hr * 100


# Fungsi 3: Capacity Factor
def capacity_factor(energi_mwh, daya_mw, jam):
    """Menghitung capacity factor dalam persen."""
    if energi_mwh <= 0:
        raise ValueError("Energi harus lebih dari nol.")
    return energi_mwh / (daya_mw * jam) * 100


# Fungsi 4: Konsumsi Spesifik
def konsumsi_spesifik(batubara_t, energi_mwh):
    """Menghitung konsumsi spesifik dalam satuan ton/MWh."""
    if energi_mwh <= 0:
        raise ValueError("Energi harus lebih dari nol.")
    return batubara_t / energi_mwh


# Fungsi 5: Status Hari
def status_hari(efisiensi, jam_operasi):
    """Menentukan status operasi berdasarkan efisiensi dan jam operasi."""
    if jam_operasi < 24:
        return "TIDAK PENUH"
    elif efisiensi < 36.3:
        return "EFISIENSI RENDAH"
    else:
        return "NORMAL"


# Pengujian fungsi menggunakan assert
assert heat_rate(1, 2, 4400) == 2200
assert round(efisiensi_termal(1, 2, 4400), 2) == 39.09
assert capacity_factor(120, 100, 24) == 5
assert konsumsi_spesifik(10, 100) == 0.1
assert status_hari(37, 24) == "NORMAL"
assert status_hari(35, 24) == "EFISIENSI RENDAH"
assert status_hari(37, 18) == "TIDAK PENUH"

print("Semua pengujian fungsi berhasil.")
```

Penjelasan:

Pada nomor ini, dibuat lima fungsi untuk menghitung dan menentukan indikator kinerja pembangkit listrik (KPI). Fungsi **heat_rate()** digunakan untuk menghitung kebutuhan energi panas dari batubara, **efisiensi_termal()** untuk menghitung efisiensi pembangkit, **capacity_factor()** untuk menghitung persentase pemanfaatan kapasitas pembangkit, dan **konsumsi_spesifik()** untuk menghitung jumlah batubara yang digunakan setiap MWh energi listrik. Selain itu, fungsi **status_hari()** digunakan untuk menentukan kondisi operasi pembangkit berdasarkan jam operasi dan nilai efisiensinya.

Data energi, batubara, dan jam operasi Unit 1 juga dimasukkan sebagai data awal yang akan digunakan dalam perhitungan. Setiap fungsi dilengkapi dengan *docstring* dan pengujian menggunakan **assert** untuk memastikan hasil perhitungan sesuai dengan nilai yang diharapkan. Penanganan **ValueError** ditambahkan untuk mencegah perhitungan ketika energi bernilai nol atau negatif.

**Nomor 2: Membuat Prosedur Cetak Laporan**

```python
# Nama: Intan Dwinancy Aprilia Nainggolan
# NIM: 4232501035

def cetak_laporan(energi, batubara, jam_operasi, daya_mw, nilai_kalor):
    """Mencetak tabel KPI pembangkit selama beberapa hari."""

    print(f"{'Hari':>4} {'HR (kcal/kWh)':>14} {'Efi (%)':>9} "
          f"{'CF (%)':>9} {'t/MWh':>9} {'Status':>20}")

    for i in range(len(energi)):
        hr = heat_rate(batubara[i], energi[i], nilai_kalor)

        efi = efisiensi_termal(
            batubara[i], energi[i], nilai_kalor
        )

        cf = capacity_factor(
            energi[i], daya_mw, jam_operasi[i]
        )

        konsumsi = konsumsi_spesifik(
            batubara[i], energi[i]
        )

        status = status_hari(efi, jam_operasi[i])

        print(f"{i + 1:>4} {hr:>14,.2f} {efi:>9.2f} "
              f"{cf:>9.2f} {konsumsi:>9.3f} {status:>20}")


# Menampilkan laporan KPI Unit 1
cetak_laporan(
    energi,
    batubara,
    jam_operasi,
    daya_unit1,
    nilai_kalor_unit1
)
```

Penjelasan:

Prosedur **cetak_laporan()** digunakan untuk menampilkan hasil perhitungan KPI Unit 1 selama 30 hari dalam bentuk tabel. Prosedur ini memanggil fungsi **heat_rate()**, **efisiensi_termal()**, **capacity_factor()**, **konsumsi_spesifik()**, dan **status_hari()** yang telah dibuat pada Nomor 1. Dengan demikian, perhitungan tidak perlu ditulis ulang karena cukup menggunakan fungsi yang sudah tersedia.

**Nomor 3: Membuat Ringkasan KPI Unit 1**

```python
# Nama: Intan Dwinancy Aprilia Nainggolan
# NIM: 4232501035

# Menghitung efisiensi setiap hari
daftar_efisiensi = []

for i in range(len(energi)):
    efi = efisiensi_termal(
        batubara[i],
        energi[i],
        nilai_kalor_unit1
    )
    daftar_efisiensi.append(efi)

# Menghitung rata-rata efisiensi
rata_efisiensi = sum(daftar_efisiensi) / len(daftar_efisiensi)

# Menghitung capacity factor bulanan
total_energi = sum(energi)
total_jam = sum(jam_operasi)

cf_bulanan = total_energi / (daya_unit1 * total_jam) * 100

# Menghitung jumlah hari berdasarkan status
jumlah_normal = 0
jumlah_efisiensi_rendah = 0
jumlah_tidak_penuh = 0

for i in range(len(energi)):
    status = status_hari(daftar_efisiensi[i], jam_operasi[i])

    if status == "NORMAL":
        jumlah_normal += 1
    elif status == "EFISIENSI RENDAH":
        jumlah_efisiensi_rendah += 1
    else:
        jumlah_tidak_penuh += 1

# Menentukan tiga hari dengan efisiensi terendah
data_efisiensi = []

for i in range(len(energi)):
    data_efisiensi.append((i + 1, daftar_efisiensi[i]))

data_terendah = sorted(data_efisiensi, key=lambda x: x[1])[:3]

# Menampilkan ringkasan KPI
print("RINGKASAN KPI UNIT 1")
print("-----------------------------")
print(f"Rata-rata efisiensi : {rata_efisiensi:.2f}%")
print(f"Capacity factor     : {cf_bulanan:.2f}%")

print("\nJumlah hari berdasarkan status:")
print(f"Normal              : {jumlah_normal} hari")
print(f"Efisiensi rendah    : {jumlah_efisiensi_rendah} hari")
print(f"Tidak penuh         : {jumlah_tidak_penuh} hari")

print("\nTiga hari dengan efisiensi terendah:")
for hari, efi in data_terendah:
    print(f"Hari ke-{hari}: {efi:.2f}%")
```

Penjelasan:

Pada nomor ini, dibuat ringkasan hasil perhitungan KPI Unit 1 selama 30 hari. Ringkasan meliputi rata-rata efisiensi termal, capacity factor bulanan, jumlah hari berdasarkan masing-masing status operasi, serta tiga hari dengan efisiensi termal terendah.

Perhitungan menggunakan fungsi yang telah dibuat pada Nomor 1. Data efisiensi dihitung untuk setiap hari, kemudian digunakan untuk memperoleh rata-rata dan menentukan tiga hari dengan efisiensi terendah. Capacity factor bulanan dihitung dari total energi yang dihasilkan dibagi dengan total kapasitas energi yang tersedia selama 30 hari, dengan memperhitungkan jam operasi setiap harinya.
