```python
# ==========================================
# Jawaban B.1 Soal Umum nomor 3
# ==========================================

# Nama   : Intan Dwinancy Aprilia Nainggolan
# NIM    : 4232501035
# Tanggal: 26 September 2026
# Tujuan : Menentukan apakah seseorang boleh mengemudi sendiri berdasarkan umur dan kepemilikan SIM.

umur = int(input("Masukkan umur: "))
punya_sim = input("Apakah memiliki SIM? (True/False): ")

punya_sim = punya_sim == "True"

cek_kelayakan = (umur >= 17) and (punya_sim == True)

boleh_mengemudi = cek_kelayakan

print(f"Boleh mengemudi sendiri: {boleh_mengemudi}")
```

```python
# ==========================================
# Jawaban B.2 Soal Konteks Energi nomor 4
# ==========================================

# Nama    : Intan Dwinancy Aprilia Nainggolan
# NIM     : 4232501035
# Tanggal : 26 September 2026
# Tujuan  : Menghitung efisiensi termal berdasarkan energi input dan energi output.

energi_input = float(input("Masukkan nilai energi input (MJ): "))
energi_output = float(input("Masukkan nilai energi output (MJ): "))

# Rumus efisiensi termal:
# efisiensi (%) = (energi output / energi input) * 100
efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal unit pembangkit: {efisiensi:.2f}%")
```
